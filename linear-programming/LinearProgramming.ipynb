```python
from pulp import LpProblem, LpMaximize, LpMinimize, LpVariable, lpSum, LpBinary, LpInteger
```

Problem 1: Product Mix (Basic LP)

```python
# 1. Initialize Model
model = LpProblem("Prob1_Product_Mix", LpMaximize)

# 2. Decision Variables (Continuous)
A = LpVariable("Product_A", lowBound=0) # Default cat is Continuous
B = LpVariable("Product_B", lowBound=0)

# 3. Objective Function (Profit)
model += 40 * A + 30 * B

# 4. Constraints
model += 2 * A + 1 * B <= 100, "Labor"
model += 3 * A + 2 * B <= 120, "Material"

# 5. Solve and Print
model.solve()
print(f"Produce A: {A.value()}")
print(f"Produce B: {B.value()}")
print(f"Total Profit: ${model.objective.value()}")
```

Problem 2: Diet Optimization (LP)

```python
# 1. Initialize Model
model = LpProblem("Prob2_Diet", LpMinimize)

# 2. Decision Variables
Rice = LpVariable("Rice", lowBound=0)
Beans = LpVariable("Beans", lowBound=0)

# 3. Objective Function (Cost)
model += 2 * Rice + 3 * Beans

# 4. Constraints
model += 4 * Rice + 8 * Beans >= 40, "Protein"
model += 200 * Rice + 150 * Beans >= 2000, "Calories"

# 5. Solve and Print
model.solve()
print(f"Rice: {Rice.value()}")
print(f"Beans: {Beans.value()}")
print(f"Min Cost: ${model.objective.value()}")
```

Problem 3: Advertising Budget (LP)

```python
# 1. Initialize Model
model = LpProblem("Prob3_Ads", LpMaximize)

# 2. Decision Variables
TV = LpVariable("TV_Ads", lowBound=0)
Online = LpVariable("Online_Ads", lowBound=0)

# 3. Objective Function (Reach)
model += 1000 * TV + 400 * Online

# 4. Constraints
model += 500 * TV + 200 * Online <= 10000, "Budget"
model += TV >= 10, "Min_TV_Ads"

# 5. Solve and Print
model.solve()
print(f"TV Ads: {TV.value()}")
print(f"Online Ads: {Online.value()}")
print(f"Total Reach: {model.objective.value()}")
```

Problem 4: Factory Opening (Integer Programming)

```python
# 1. Initialize Model
model = LpProblem("Prob4_Factory", LpMaximize)

# 2. Decision Variables (Binary: 0 or 1)
Fac_A = LpVariable("Factory_A", cat="Binary")
Fac_B = LpVariable("Factory_B", cat="Binary")

# 3. Objective Function (Profit)
model += 80000 * Fac_A + 60000 * Fac_B

# 4. Constraints
model += 50000 * Fac_A + 40000 * Fac_B <= 70000, "Budget"

# 5. Solve and Print
model.solve()
print(f"Open Factory A: {Fac_A.value()}")
print(f"Open Factory B: {Fac_B.value()}")
print(f"Total Profit: ${model.objective.value()}")
```

Problem 5: Production with Fixed Cost (MIP)

```python
# 1. Initialize Model
model = LpProblem("Prob5_FixedCost", LpMaximize)

# 2. Decision Variables
# 'x' is units produced (Continuous or Integer, usually Integer for units)
x = LpVariable("Units", lowBound=0, cat="Integer")
# 'y' is the machine switch (Binary: 1=On, 0=Off)
y = LpVariable("Machine_On", cat="Binary")

# 3. Objective Function
# Profit = (Revenue * x) - (Fixed_Cost * y)
model += 10 * x - 3000 * y

# 4. Constraints
# Standard max production capacity
model += x <= 1000

# Linking Constraint (Big M method):
# If y=0, x must be 0. If y=1, x can be up to 1000.
model += x <= 1000 * y

# 5. Solve and Print
model.solve()
print(f"Machine On: {y.value()}")
print(f"Units Produced: {x.value()}")
print(f"Net Profit: ${model.objective.value()}")
```
