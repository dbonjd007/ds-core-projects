```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# User-item Ratings Matrix
ratings = pd.DataFrame(
    {
        'User': ['Alice','Bob','Charlie','David', 'Eve'],
        'Toy Story': [5,4,0,0,3],
        'Titanic': [0,5,4,0,2],
        'Avatar': [4,0,5,3,0],
        'Die Hard': [0,0,4,5,0],
        'Shutter Island': [3,0,0,4,5]
    }
).set_index('User')

ratings
```

```python
def collaborative_recommendations(name, num_recommendation):
  # Find similar users
  user_similarity = cosine_similarity(ratings.fillna(0))
  user_similarity_df = pd.DataFrame(user_similarity, index=ratings.index, columns=ratings.index)

  # Get Similar Users
  similar_users = user_similarity_df[name].sort_values(ascending=False)[1:]

  # Find movies similars to others
  user_ratings = ratings.loc[name]
  recommendations = {}

  for movie in ratings.columns:
    if user_ratings[movie] == 0: # User hasn't seen this movie
      weighted_sum = 0
      similarity_sum = 0

      for similar_user in similar_users.index[:3]: # Grab top 3 similar users
        if ratings.loc[similar_user, movie] > 0:
          similarity = user_similarity_df.loc[name, similar_user]
          weighted_sum += ratings.loc[similar_user, movie] * similarity
          similarity_sum += similarity
        if similarity_sum > 0:
          predicted_rating = weighted_sum / similarity_sum
          recommendations[movie] = predicted_rating
  print(f"Debug: Recommendations before sorting: {recommendations}") # Debug statement
  sorted_recs = sorted(recommendations.items(), key=lambda x:x[1], reverse=True)
  return [movie for movie, x in sorted_recs[:num_recommendation]]
```

```python
print("\nCollaborative Recommendations:")
print(f"For Alice, recommend: {collaborative_recommendations("Alice", 3)}")
```
