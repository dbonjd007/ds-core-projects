```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# User-item Ratings Matrix
books = pd.read_csv('Books.csv')
users = pd.read_csv('Users.csv', converters={3: pd.to_numeric})
```

```python
def collaborative_recommendations(name, num_recommendation):
  # Find similar users
  user_similarity = cosine_similarity(users.fillna(0))
  user_similarity_df = pd.DataFrame(user_similarity, index=users.index, columns=users.index)

  # Get Similar Users
  similar_users = user_similarity_df[name].sort_values(ascending=False)[1:]

  # Find users similars to others
  user_ratings = users.loc[name]
  recommendations = []

  for movie in users.columns:
    if user_ratings[movie] == 0: # User hasn't read this book
      weighted_sum = 0
      similarity_sum = 0

      for similar_user in similar_users.index[:3]: # Grab top 3 similar users
        if users.loc[similar_user, movie] > 0:
          similarity = user_similarity_df.loc[name, similar_user]
          weighted_sum += users.loc[similar_user, movie] * similarity
          similarity_sum += similarity
        if similarity_sum > 0:
          predicted_rating = weighted_sum / similarity_sum
          recommendations[movie] = predicted_rating
  print(f"Debug: Recommendations before sorting: {recommendations}") # Debug statement
  sorted_recs = sorted(recommendations.items(), key=lambda x:x[1], reverse=True)
  return [movie for movie, x in sorted_recs[:num_recommendation]]
```

```python
print("\nCollaborative Recommendations:")
print(f"For Alice, recommend: {collaborative_recommendations("Alice", 3)}")
```
