```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
import pandas as pd

# sample

x = np.array([100,200,300,400,500]).reshape (-1,1) # Advertising Spending
y = np.array([25,45,65,85,105])  #Sale

print(x)
print(y)
```

```python
# Linear Regression module
model = LinearRegression()
model.fit(x,y)
```

```python
prediction = model.predict([[y]])
print(prediction)
```

```python
plt.scatter(x,y, color="blue", label="Actual Sales")
# Calculate predictions for all x values in the original dataset
plt.plot(x, prediction, color="red", label="Model Prediction")
plt.xlabel("Advertising Spending")
plt.ylabel("Sales")
plt.title(" Simple Linear Regression")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# sample

x = np.array([75,200,300,400,500]).reshape (-1,1) # Advertising Spending
y = np.random.randint(0, 100, size=len(x))  #Sale with random whole numbers

print(x)
print(y)

model = LinearRegression()
model.fit(x,y)

prediction_y = model.predict(x) # Calculate predictions using the current x values

plt.scatter(x,y, color="blue", label="Orders Made")
plt.plot(x,prediction_y, color="red", label="Model Prediction")
plt.xlabel("Vouchers Used")
plt.ylabel("Orders")
plt.title("Product X throughout November")
plt.legend()
plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures

x = np.array([1,2,3,4,5]).reshape(-1,1) # Year
y = np.array([3,6,19,45,85])            #Population
```

```python
#Polynomial Features
poly = PolynomialFeatures()
x_poly = poly.fit_transform(x)
x_poly
```

```python
# Model

model = LinearRegression()
model.fit(x_poly,y)
predict = model.predict(x_poly)
```

```python
# Plot
plt.scatter(x,y, color="blue")
plt.plot(x,predict, color="red")
plt.xlabel("Year")
plt.ylabel("Population")
plt.title("Polynomical Regression")
plt.show()
```
